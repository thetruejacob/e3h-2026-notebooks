## Part 0: Set-up

In this problem set you will analyze **life expectancy gradients** for the United States. Before starting, please review the slide deck `E3H_8d_coefficient.pdf` on Canvas.

The data contains four sets of life expectancy variables, each computed with a different method:

- `LE_CensusTract_2015` and `LE_CensusTract_2021` — life expectancy at birth by the income percentile of an individual's **census tract** in California (2015 and 2021). Based on *Schwandt et al. (2022, JAMA)*.
- `LE_County_2015` — life expectancy at birth by the income percentile of an individual's **county** in California in 2015. This method has been used e.g. in *Currie & Schwandt (2016, Science)*.
- `LE_Individual_Income_2014_age40` — life expectancy at **age 40** by **individual household income** percentile in 2014. From *Chetty et al. (2016, JAMA)*.

`percentile` is the income percentile of the county / census tract / individual income group.

## Figure 1: County vs census tract gradients (2015)

The county-based gradient has been used in many studies to measure inequality in mortality, but it has been criticized for overlooking inequality *within* counties.

```python
plt.figure(figsize=(8, 5))
plt.plot(le.percentile, le.LE_County_2015, label="LE by county")
plt.plot(le.percentile, le.LE_CensusTract_2015, label="LE by census tract")
plt.xlabel("income percentile"); plt.ylabel("life expectancy")
plt.legend()
plt.show()
```
